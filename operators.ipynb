```python
#arithemetic operators
#addition 
a=5
b=10
c=a+b
print(c)

#subtraction
a=7
b=3
c=a-b
print(c)

#multipication
a=6
b=4
c=a*b
print(c)

#division
a=4
b=2
c=a/b
print(c)

#floor division
a=6
b=4
c=a//b
print(c)

#modulo division
a=7
b=3
c=a%b
print(c)

#exponential
a=5
b=3
c=a**b
print(c)
```

```python
#assignment operators
#add and assign(+=)
a=5
b+=3
print(b)

#subtract and assign(-=)
a=8
b-=2
print(b)

#mulitiplicate and assign(*=)
a=6
b*=2
print(b)

#exponential(**=)
a=5
b**=2
print(b)

#divide and assign(/=)
a=6
b/=2
print(b)

#floor division(//=)
a=5
b//=3
print(b)

#modulo division(%=)
a=5
b%=2
print(b)
```

```python
#comparison operators
#equal to (==)
a=5
b=10
print(a==b)

#not equal to (!=)
a=5
b=10
print(a!=b)

#greater than (>)
a=5
b=10
print(a>b)

#lesser than (<)
a=5
b=10
print(a<b)
```

```python
#logical operators
#and operator
a=True
b=True
print(a and b)

a=True
b=False
print(a and b)

a=False
b=False
print(a and b)

a=False
b=True
print(a and b)


#or operator
a=True
b=True
print(a or b)

a=True
b=False
print(a or b)

a=False
b=False
print(a or b)

a=False
b=True
print(a or b)

#not operators
a=True
b=True
print(not a)

a=False
b=False 
print(not a)




```

```python
#membership operators
fruits=['apple','oranges','grapes','watermelon']
print('apple' in fruits)
print('grapes' not in fruits)
print('avacado' in fruits)
print('avacado' not in fruits)

name='lasya'
print('l' in name)
print('l'not in name)
print('b'in name)
print('b'not in name)
```

```python
#bitwise operators
#and operator(&)
a=7
b=4
print(a&b)

#or operator(|)
a=7
b=4
print(a | b)

#xor operator(^)
a=7
b=4
print(a ^ b)

#left shift operator(<<)
a=7
b=4
print(a<<b)

#right shift operator(>>)
a=7
b=4
print(a>>b)
```

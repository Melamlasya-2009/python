```python
#for loop
#used to repeat code or iterate through a sequence.

for i in range (1,6):
    print(i)
```

```python
#use while when repetition depends on a condition.
#Looping through numbers 1 to 5 using while loop
i=1
while i<=5:
    print(i)
    i=i+1
```

```python
#multiplication table
number=int(input("enter number:"))
for i in range(1,11):
    print(number,"x",i,"=",number*i)
```

```python
#sum of numbers from 1to n
n=int(input("enter the number:"))

total=0

for i in range(1,n+1):
    total=total+i
    
print("sum:",total)
```

```python
#factorial of a number
n=int(input("enter number:"))

factorial=1

for i in range(1,n+1):
    factorial=factorial*i
print("factorial:",factorial)
    
```

```python
#sum of even numbers from 2 to n
n=int(input("enter n:"))

total=0

for i in range(2,n+1,2):
    total=total+i
print("sum:",total)
```

```python
#count of multiples of 3
n=int(input("enter n:"))

count=0

for i in range(1,n+1):
    if i%3==0:
        count=count+1
print("count:",count)
```

```python
#sum of multiples of 5
n=int(input("enter n:"))

total=0
for i in range(1,n+1):
    if i%5==0:
        total=total+i
print("sum:",total)
```

```python
#print total of numbers entered by user until 0 is entered
total=0

number=int(input("enter the number:"))

while number !=0:
    total=total+number
    number=int(input("enter a number:"))

print("total:",total)
```

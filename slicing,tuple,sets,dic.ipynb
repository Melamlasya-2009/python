```python
#slicing

numbers=[10,20,30,40,50,60,70,80]

print(numbers[::-1])
```

```python
#slicing

numbers=[10,20,30,40,50,60,70,80]

print(numbers[-1:2:-3])
```

```python
#tuples
student=("lasya",17,"python")

print(student[0])
```

```python
#access values in a tuple
student=("lasya",17,"python")

print(student[0])
print(student[1])
print(student[2])
```

```python
#immutable nature of tuples
student=("lasya",17,85.5)

student[1]=22
```

```python
subjects={"python","java","python","java script"}

print(subjects)
```

```python
#add values to a set
subjects={"python","java"}

subjects.add("SQL")

print(subjects)
```

```python
#remove values from a set 
subjects.remove("java")

print(subjects)
```

```python
#sets do not allow duplicate values
numbers={1,2,2,3,3,4}

print(numbers)
```

```python
#dictionaries
student={
    "name":"Lasya",
    "age":17,
    "course":"python"
}

print(student)
```

```python
#access elements in disc
print(student["name"])
print(student["age"])
print(student["course"])
```

```python
#add new data to a dictionary
student["city"]="Tirupati"

print(student)
```

```python
#

print(student.keys())

print(student.values())

print(student.items())

print(student.get("name"))

student.update({"age":18})

print(student.pop("age"))

student.popitem()

print(student)

student.setdefault("age",18)

print(student)

```

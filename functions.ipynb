```python
#function for even/odd
def check_even_odd(number):
    if number%2==0:
        return "even"
    else:
        return "odd"
result=check_even_odd(6)
print(result)
```

```python
#pass or fail
def check_marks(marks):
    if marks>=45:
        return "pass"
    else:
        return "fail"
result=check_marks(60)
print(result)
```

```python
#function for largest of two numbers
def largest(a,b):
    if a>b:
        return a
    else:
        return b
result=largest(10,20)

print("largest:",result)
```

```python
#age eligibility
def check_eligibility(age):
    if age>=18:
        return "eligible"
    else:
        return "not eligible"
result=check_eligibility(45)
print(result)
```

```python
#temparature check
def check_temparature(degrees):
    if degrees>=45:
        return "high temparature"
    else:
        return "normal temparature"
result=check_temparature(67)
print(result)
```

```python
#function + user input
def square(number):
    return number*number
number=int(input("enter number:"))
result=square(number)
print("square:",result)
```

```python
#function + loop
def multiplication_table(number):
    for i in range(1,11):
        print(number,"x",i,"=",number*i)
number=int(input("enter number:"))
multiplication_table(number)
```

```python
#largest of three numbers
def largest(a,b,c):
    if a>=b and a>=c:
        return a
    elif b>=a and b>=c:
        return b
    else:
        return c
a=int(input("enter a:"))
b=int(input("enter b:"))
c=int(input("enter c:"))
print("largest=",largest(a,b,c))
```

```python
#largest of two numbers
def largest(a,b):
    if a>=b:
        return a
    else:
        return b
a=int(input("enter a:"))
b=int(input("enter b:"))
print("largest=",largest(a,b))
```

```python
#GRADE calculation
def find_grade(marks):
    if marks>=95:
        return "A*"
    if marks>=90:
        return "A"
    if marks>=80:
        return "B"
    if marks>=70:
        return "C"
    if marks>=60:
        return "D"
    if marks>=40:
        return "E"
    else:
        return "fail"

marks=int(input("enter marks:"))
print("Grade=",find_grade(marks))
```
